#### Write a program using Python to implement Self-Organizing Map (SOM).

```python
# Self Organizing Map (SOM)

from minisom import MiniSom
import numpy as np
```

```python
# Sample data
data = np.array([
    [1, 2],
    [2, 3],
    [3, 4],
    [8, 7],
    [9, 8],
    [10, 9]
])

# Create SOM
som = MiniSom(3, 3, 2, sigma=1.0, learning_rate=0.5)

som.random_weights_init(data)
som.train_random(data, 100)

print("SOM Training Completed")

for d in data:
    print(d, "->", som.winner(d))
```
