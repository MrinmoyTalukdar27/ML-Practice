#### Write a program implement Backpropagation Algorithm.

```python
import numpy as np
from sklearn.neural_network import MLPClassifier

# Backpropagation using sklearn's MLPClassifier
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

y = np.array([0, 1, 1, 0])

model = MLPClassifier(hidden_layer_sizes=(4,),
                      activation='logistic',
                      solver='sgd',
                      learning_rate_init=0.1,
                      max_iter=10000,
                      random_state=1)

model.fit(X, y)

predictions = model.predict(X)
probabilities = model.predict_proba(X)[:, 1]

print('=' * 50)
print('Input data:')
print(X)
print('=' * 50)
print('Predictions:')
print(predictions)
print('Prediction probabilities:')
print(np.round(probabilities, 4))
print('Training accuracy:')
print(model.score(X, y))
print('=' * 50)
```
