#### Write a program using Python to implement Logistic Regression with Multiple Variables

```python
# Logistic Regression with Multiple Variables
from sklearn.linear_model import LogisticRegression
import numpy as np
```

```python
# Features: Age, Salary
X = np.array([
    [22, 25000],
    [25, 30000],
    [47, 50000],
    [52, 60000],
    [46, 55000]
])
```

```python
# Purchased (0 = No, 1 = Yes)
y = np.array([0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

# Prediction
new_data = np.array([[40, 45000]])
prediction = model.predict(new_data)

print("Prediction:", prediction[0])
```
