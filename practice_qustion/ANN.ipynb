#### Write a program using Python to implement Artificial Neural Network.

```python
# Artificial Neural Network
from sklearn.neural_network import MLPClassifier
```

```python
#Training data
X = [[0,0], [0,1], [1,0], [1,1]]
y = [0, 1, 1, 0]
```

```python
model = MLPClassifier(
    hidden_layer_sizes=(4,),
    max_iter=2000,
    random_state=1
)

model.fit(X, y)

print("Predictions:")
print(model.predict(X))
```
