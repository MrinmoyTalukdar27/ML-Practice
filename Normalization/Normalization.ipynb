###  Now we can see how to use the min-max scaler

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df =pd.read_csv("Wine dataset.csv",header=0,usecols=[0,1,2])
df.columns=['class', 'Alcohol' , 'Malic acid']
df.head(5)
# Here we are using only three columns for our normalization
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.drop('class',axis=1),df['class'],test_size=0.3,random_state=0)
```

```python
from sklearn.preprocessing import MinMaxScaler
# xi'=(xi-x_min)/(x_max-x_min)
# xi' is the value of the trensform column
# xi value to be transform
# x_min & x_max are the min and max value of the column

# The above formula  is used  minmax scaler for normalization 
scaler=MinMaxScaler()
scaler.fit(x_train)

x_train_scaled=scaler.transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
x_train_scaled=pd.DataFrame(x_train_scaled,columns=x_train.columns)
X_test_scaled=pd.DataFrame(x_test_scaled,columns=x_test.columns)
```

```python
x_train_scaled.head(10)
```

```python
np.round(x_train.describe())
```

```python
np.round(x_train_scaled.describe())
#Here after scaling we can see that min value of the dataset is 0 & the maximum value of the dataset is 1
# It scaled down the values between 0,1 & it would not chane the actual shape or qualites of the datset
```

```python
fig, (ax1,ax2)= plt.subplots(ncols=2,figsize=(12,5))
ax1.scatter(x_train["Alcohol"],x_train['Malic acid'],c=y_train)
ax1.set_title('Before sacling')

ax2.scatter(x_train_scaled["Alcohol"],x_train_scaled['Malic acid'],c=y_train)
ax2.set_title('After scalling')
plt.show()
# here we can see that the nature of the actual data remain same after and befor scalling
# it just scaled the value for us so that it become easy to perform operation
```
