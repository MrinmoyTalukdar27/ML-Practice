```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('housing.csv',usecols=['FireplaceQu','GarageQual','SalePrice'])
```

```python
df.head(5)
```

```python
df.isnull().mean()*100
```

```python
# now we will work on column Garagequality('GarageQual')
# where our 5.5 parsent data is missing
df['GarageQual'].value_counts().plot(kind='bar',color='yellow')
plt.xlabel('GarageQuality')
plt.ylabel('Numer of houses')
```

```python
# now we are checking which value is most frequent
df['GarageQual'].mode()
```

```python
fig, ax=plt.subplots()
df[df['GarageQual']=='TA']['SalePrice'].plot(kind='kde',ax=ax,color='yellow')
df[df['GarageQual'].isnull()]['SalePrice'].plot(kind='kde',ax=ax,color='red')

lines,labels=ax.get_legend_handles_labels()
labels=['Houses with TA value','Houses with no value']
ax.legend(lines,labels,loc='best')

plt.title('GarageQual')
plt.show()
```

```python
temp=df[df['GarageQual']=='TA']['SalePrice']
```

```python
df['GarageQual'].fillna('TA',inplace=True)
```

```python
df['GarageQual'].value_counts().plot(kind='bar',color='yellow')
plt.show()
```

```python
fig, ax=plt.subplots()
temp.plot(kind='kde',ax=ax)

df[df['GarageQual']=='TA']['SalePrice'].plot(kind='kde',ax=ax,color='red')

lines,labels=ax.get_legend_handles_labels()
labels=['Houses with TA value','Houses with no value']
ax.legend(lines,labels,loc='best')

plt.title('GarageQual')
```

```python
# now we will remove missing values from the fireplace column
df['FireplaceQu'].value_counts().plot(kind='bar',color='yellow')
plt.show()
```

```python
# now we need the most frequent value in the column
df['FireplaceQu'].mode()
```

```python
# Now we see the data distrubution between missing value and Gd value 
fig, ax=plt.subplots()

df[df['FireplaceQu']=='Gd']['SalePrice'].plot(kind='kde',color='yellow')
df[df['FireplaceQu'].isnull()]['SalePrice'].plot(kind='kde',color='red')

lines,labels=ax.get_legend_handles_labels()
labels=['House with Gd','House with null values']

ax.legend(lines,labels,loc='best')
plt.title('FireplaceQu')
plt.show()
```
