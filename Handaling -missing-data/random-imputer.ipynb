```python
#libary we need
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# sklearn function
from sklearn .model_selection import train_test_split
```

```python
df=pd.read_csv('modified-titanic.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head(5)
```

```python
#now we will check the missing values present in our data in persentage
df.isnull().mean()*100
```

```python
# now herer we wre creating our x and y
x=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
#Now we we will do the train test split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
x_train['Age_impu'] = x_train['Age'] # here we are creating a temp column to see the changes befor and after the missing the imputation.
x_test['Age_impu'] = x_test['Age']

x_train['Age_impu'][x_train['Age_impu'].isnull()] = x_train['Age'].dropna().sample(x_train['Age_impu'].isnull().sum()).values
x_test['Age_impu'][x_test['Age_impu'].isnull()] = x_test['Age'].dropna().sample(x_test['Age_impu'].isnull().sum()).values
```

```python
x_train
```

```python
# now we see the data distrubution befor and after the imputation
sns.displot(x_train['Age'],label='Before imputation',color='green')
sns.displot(x_train['Age_impu'],label='After the imputation',color='orange')
plt.legend()
plt.show()
```
