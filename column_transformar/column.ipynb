```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv('covid_toy.csv')
```

```python

df.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.drop('has_covid',axis=1),df['has_covid'],test_size=0.20,random_state=42)
# here we are spliting the traning and testing data in two parts
```

```python
x_train.head(10)
```

```python

from sklearn.compose import ColumnTransformer
# here we are calling the columntransformar class from sklearn
```

```python
transform=ColumnTransformer(transformers=[
    ('tran1',SimpleImputer(),[' fever ']),
    # here if we need to pass transformar we need to passs them through an tupplr like tran1,tran2,tran3
    # next inside the tupple we have to pass the transformar object like SimpleImputyer
    # next we have to pass the column names where we want to use this transformar and the column names should be pass through in a  list
    # like ['fever]

    ('tran2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    # we have to use three diffrent tranfromar so we are passing three tupple
    # next inside the tupple we have to pass the transformar object like ordinaEncoding
    # after that we have to pass the catagories inside the ordinalEncoder,like[['mild','stong']]
     # next we have to pass the column names where we want to use this transformar and the column names should be pass through in a  list

    ('tran3',OneHotEncoder(sparse_output= False,drop='first'),['gender','city'])
    # same as simple imputer
    #sparse_output =False is use so that we dont get sparse array ,drop='first' is mainly used to handle multicollinearity

],remainder='passthrough')#reminder=passthrough mean the column we are not transforming those are pass through without any changes 
```

```python
x_train_transform=transform.fit_transform(x_train)
X_test_transform=transform.fit_transform(x_test)
# here we fit the trainning and testing data
```
