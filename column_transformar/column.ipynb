```python
import pandas as pd
import numpy as np
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv('covid_toy.csv')
```

```python

df
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop('has_covid',axis=1),df['has_covid'],test_size=0.20,random_state=42)
```

```python
x_train.head()
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transform=ColumnTransformer(transformers=[
    ('tran1',SimpleImputer(),['fever']),
    ('tran2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tran3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
x_train_transform=transform.fit_transform(x_train)
```

```python
X_test_transform=transform.fit_transform(x_test).shape
```

```python
x_train_transform=pd.DataFrame(x_train_transform,columns=x_train.columns)
X_test_transform=pd.DataFrame(X_test_transform,columns=x_test.columns)
```
